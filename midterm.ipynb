```python
from datetime import datetime
from datetime import date, timedelta
import re

# Define the function "calculate_days"
def calculate_days(string): 
    time1=re.search(r"Aired: ([^-]+) -", string).group(1)
    time2=re.search(r"- (.+)", string).group(1)
    days=datetime.strptime(time1, "%b %d, %Y") - datetime.strptime(time2, "%b %d, %Y")
    return abs(days)
```

```python
phrase = 'Aired: Jun 18, 2022 - Aug 28, 2022'
# test that your function works

calculate_days(phrase)
```
